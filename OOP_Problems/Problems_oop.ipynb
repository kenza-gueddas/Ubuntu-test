## PROBLEM 1 ##

```python
class Complex:
    def __init__(self, real=0.0, imag=0.0):
        self.real = float(real)
        self.imag = float(imag)

    def __add__(self, other):
        if not isinstance(other, Complex):
            return NotImplemented
        return Complex(self.real + other.real, self.imag + other.imag)

    def __sub__(self, other):
        if not isinstance(other, Complex):
            return NotImplemented
        return Complex(self.real - other.real, self.imag - other.imag)

    def __repr__(self):
        sign = "+" if self.imag >= 0 else "-"
        return f"{self.real} {sign} {abs(self.imag)}i"


c1 = Complex(2, 3)
c2 = Complex(4, -5)

print("c1 =", c1)
print("c2 =", c2)
print("c1 + c2 =", c1 + c2)
print("c1 - c2 =", c1 - c2)

```

## PROBLEM 2 ##

```python
class Person:
    def __init__(self, name, age, gender):
        self.name = name
        self.age = int(age)
        self.gender = gender

    def __gt__(self, other):
        if not isinstance(other, Person):
            return NotImplemented
        return self.age > other.age

    def __repr__(self):
        return f"Person(name={self.name!r}, age={self.age}, gender={self.gender!r})"


# Tests
p1 = Person("Kenza", 19, "F")
p2 = Person("YASSINE", 19, "M")

print("p1 =", p1)
print("p2 =", p2)
print("p1 > p2 ?", p1 > p2)
print("p2 > p1 ?", p2 > p1)
```

### PROBLEM 4 ###

```python
class Vector:
    def __init__(self, x, y, z):
        self.x = float(x)
        self.y = float(y)
        self.z = float(z)

    def __mul__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        return self.x * other.x + self.y * other.y + self.z * other.z

    def __repr__(self):
        return f"Vector(x={self.x}, y={self.y}, z={self.z})"


# Tests / prints
v1 = Vector(1, 2, 3)
v2 = Vector(4, -1, 2)

print("v1 =", v1)
print("v2 =", v2)
print("v1 * v2 ) =", v1 * v2)
```

### PROBLEM 5 ###

```python
class Circle:
    def __init__(self, radius):
        self.radius = float(radius)

    def _key(self):
        return self.radius

    def __lt__(self, other):
        if not isinstance(other, Circle):
            return NotImplemented
        return self._key() < other._key()

    def __le__(self, other):
        if not isinstance(other, Circle):
            return NotImplemented
        return self._key() <= other._key()

    def __gt__(self, other):
        if not isinstance(other, Circle):
            return NotImplemented
        return self._key() > other._key()

    def __ge__(self, other):
        if not isinstance(other, Circle):
            return NotImplemented
        return self._key() >= other._key()

    def __eq__(self, other):
        if not isinstance(other, Circle):
            return NotImplemented
        return self._key() == other._key()

    def __ne__(self, other):
        if not isinstance(other, Circle):
            return NotImplemented
        return self._key() != other._key()

    def __repr__(self):
        return f"Circle(radius={self.radius})"


# Test
c1 = Circle(3)
c2 = Circle(5)
c3 = Circle(3)

print("c1 =", c1, " c2 =", c2, " c3 =", c3)
print("c1 < c2 :", c1 < c2)
print("c1 <= c3:", c1 <= c3)
print("c2 > c1 :", c2 > c1)
print("c2 >= c3:", c2 >= c3)
print("c1 == c3:", c1 == c3)
print("c1 != c2:", c1 != c2)
```

### PROBLEM 6 ###

```python
import math

class Point2D:
    def __init__(self, x, y):
        self.x = float(x)
        self.y = float(y)

    def distance_from_origin(self):
        return math.sqrt(self.x**2 + self.y**2)

    def __lt__(self, other):
        if not isinstance(other, Point2D):
            return NotImplemented
        return self.distance_from_origin() < other.distance_from_origin()

    def __le__(self, other):
        if not isinstance(other, Point2D):
            return NotImplemented
        return self.distance_from_origin() <= other.distance_from_origin()

    def __gt__(self, other):
        if not isinstance(other, Point2D):
            return NotImplemented
        return self.distance_from_origin() > other.distance_from_origin()

    def __ge__(self, other):
        if not isinstance(other, Point2D):
            return NotImplemented
        return self.distance_from_origin() >= other.distance_from_origin()

    def __eq__(self, other):
        if not isinstance(other, Point2D):
            return NotImplemented
        return self.distance_from_origin() == other.distance_from_origin()

    def __ne__(self, other):
        if not isinstance(other, Point2D):
            return NotImplemented
        return self.distance_from_origin() != other.distance_from_origin()

    def __repr__(self):
        d = self.distance_from_origin()
        return f"Point2D(x={self.x}, y={self.y}, dist={d:.3f})"


# Test
p1 = Point2D(3, 4)     # dist = 5
p2 = Point2D(1, 1)     # dist ~ 1.414
p3 = Point2D(-3, -4)   # dist = 5

print("p1 =", p1)
print("p2 =", p2)
print("p3 =", p3)

print("p2 < p1 :", p2 < p1)
print("p1 == p3:", p1 == p3)
print("p1 > p2 :", p1 > p2)
print("p1 != p2:", p1 != p2)
```

### PROBLEM 7 ###

```python
class Matrix:
    def __init__(self, data):
        # data: list of lists
        if not data or not all(isinstance(row, list) and row for row in data):
            raise ValueError("Matrix data must be a non-empty list of non-empty rows")

        row_len = len(data[0])
        if any(len(row) != row_len for row in data):
            raise ValueError("All rows must have the same length")

        # Convert to floats for safety
        self.data = [[float(x) for x in row] for row in data]
        self.rows = len(self.data)
        self.cols = len(self.data[0])

    def __mul__(self, other):
        if not isinstance(other, Matrix):
            return NotImplemented
        if self.cols != other.rows:
            raise ValueError(f"Incompatible sizes: {self.rows}x{self.cols} cannot multiply {other.rows}x{other.cols}")

        result = []
        for i in range(self.rows):
            row_result = []
            for j in range(other.cols):
                s = 0.0
                for k in range(self.cols):
                    s += self.data[i][k] * other.data[k][j]
                row_result.append(s)
            result.append(row_result)

        return Matrix(result)

    def __repr__(self):
        return "Matrix(" + ",\n       ".join(str(row) for row in self.data) + ")"


# Tests / prints
A = Matrix([[1, 2, 3],
            [4, 5, 6]])

B = Matrix([[7, 8],
            [9, 10],
            [11, 12]])

print("A =", A)
print("B =", B)
print("A * B =", A * B)
```

### Inheritance — Problem 1 ###

```python
class Vehicle:
    def drive(self):
        return "Driving a vehicle"


class Car(Vehicle):
    def drive(self):
        return "Driving a car"


class Bicycle(Vehicle):
    def drive(self):
        return "Riding a bicycle"


# Test
v = Vehicle()
c = Car()
b = Bicycle()

print(v.drive())
print(c.drive())
print(b.drive())
```

### PROBLEM 2 ###

```python
class Person:
    def __init__(self, name, age, address):
        self.name = name
        self.age = age
        self.address = address

    def introduce(self):
        return f"My name is {self.name}, I am {self.age} years old and I live at {self.address}."


class Student(Person):
    def __init__(self, name, age, address, field_of_study):
        super().__init__(name, age, address)
        self.field_of_study = field_of_study

    def introduce(self):
        return (
            f"My name is {self.name}, I am {self.age} years old, "
            f"I study {self.field_of_study} and I live at {self.address}."
        )


class Employee(Person):
    def __init__(self, name, age, address, company):
        super().__init__(name, age, address)
        self.company = company

    def introduce(self):
        return (
            f"My name is {self.name}, I am {self.age} years old, "
            f"I work at {self.company} and I live at {self.address}."
        )


# Test
p = Person("Mehdi", 50, "Tunis")
s = Student("Kenza", 19, "Cergy", "Computer Science")
e = Employee("Ines", 30, "Nex York", "Google")

print(p.introduce())
print(s.introduce())
print(e.introduce())
```

### PROBLEM 3 ###

```python
class BankAccount:
    def __init__(self, account_number, balance=0.0):
        self.account_number = account_number
        self.balance = float(balance)

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            return True
        return False

    def transfer(self, amount, destination_account):
        if self.withdraw(amount):
            destination_account.deposit(amount)
            return True
        return False


class SavingsAccount(BankAccount):
    def __init__(self, account_number, balance=0.0, interest_rate=0.0):
        super().__init__(account_number, balance)
        self.interest_rate = float(interest_rate)

    def add_interest(self):
        self.balance += self.balance * self.interest_rate

    def check_balance(self):
        return self.balance

# Test
acc1 = BankAccount("ACC001", 1000)
acc2 = SavingsAccount("ACC002", 500, 0.05)

print("Initial balances:")
print("acc1:", acc1.balance)
print("acc2:", acc2.balance)

print("\nTransfer 300 from acc1 to acc2:")
success = acc1.transfer(300, acc2)
print("Transfer success:", success)

print("acc1:", acc1.balance)
print("acc2:", acc2.balance)

print("\nAdd interest to acc2:")
acc2.add_interest()
print("acc2 after interest:", acc2.check_balance())
```
